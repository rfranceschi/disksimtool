```python
from pathlib import Path

import gofish as gf
from matplotlib import pyplot as plt
from matplotlib.colors import LogNorm, Normalize

RUNS_DIR = Path('runs')
```

```python
runs_to_check = RUNS_DIR / 'runs_test_T/'

for _run in runs_to_check.iterdir():
    for _fits_file in _run.glob('*.fits'):
        f, ax = plt.subplots()

        icube = gf.imagecube(_fits_file)
        norm = LogNorm(1e-3 * icube.data.max(), 1e0 * icube.data.max())
        im = ax.imshow(icube.data, extent=icube.extent, norm=norm)
        cb = plt.colorbar(im, ax=ax, fraction=0.046, pad=0.04)

        ax.set_xlim(-0.5, 0.5)
        ax.set_ylim(-0.5, 0.5)
        name = _fits_file.stem
        ax.set_title('/'.join([_fits_file.parent.name[:15], _fits_file.name]))
        plt.show()
```
